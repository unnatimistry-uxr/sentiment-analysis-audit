#### ⚙️ Script Blueprint & Data Routing (Frequency Validation)
* **Execution Logic:** 
  1. Compiles the full text of all 5-star reviews and passes it through an automated tokenization process.
  2. Pulls down and filters out standard English grammatical filler using an `nltk` stop-word exclusion array.
  3. Uses a `Counter` dictionary framework to compute exact raw string repetitions and extracts the top 50 dominant words.
* **Output Format & Destination:** A physical HTML data sheet tracking macro-trends saved to `paths.REPORTS_POS_FREQ_VALIDATION_HTML`, supplemented by an in-notebook dataframe block displaying the top 20 validated terms for localized verification.

```python
import pandas as pd
from collections import Counter
import nltk
from nltk.corpus import stopwords

# Ensure stopwords are ready
nltk.download('stopwords', quiet=True)
stop_words = set(stopwords.words('english'))

# 1. Get all 5-star reviews (Full Scale)
full_pos_reviews = df[df['score'] == 5]['cleaned_content'].dropna()

# 2. Extract and clean all words
all_words = " ".join(full_pos_reviews).lower().split()
filtered_words = [w for w in all_words if w not in stop_words and len(w) > 3]

# 3. Get Top 50 most frequent words
word_freq = Counter(filtered_words).most_common(50)
word_freq_df = pd.DataFrame(word_freq, columns=['Word', 'Frequency'])

# 4. EXPORT: Save this to help build your Codebook
word_freq_df.to_html(paths.REPORTS_POS_FREQ_VALIDATION_HTML, index=False)

print("✅ Success: frequency_validation.html saved.")
print("Take a look at the top 20 words below. Do they match your themes?")

# 5. PREVIEW: Check the top 20
word_freq_df.head(20)
```

### Step 4: The Priority Waterfall Engine: Algorithmic Theme Classification

* **What we are doing:** Deploying a rule-based Natural Language Processing (NLP) classification function across the complete 5-star review database to programmatically categorize every review into one of ten mutually exclusive business and user-experience themes.
* **Why we are doing this:** Scaled qualitative data cannot be hand-coded efficiently at production volumes. By implementing a standardized keyword heuristic, we eliminate manual bottlenecking while ensuring every piece of user praise is bucketed into actionable categories—such as tracking competitive wins or feature requests—giving product managers objective quantitative charts.
* **How we are doing this:** We construct an explicit, conditional prioritization hierarchy (The Waterfall Strategy) utilizing a string-matching heuristic function. Because a single review can mention multiple features, the data "falls" through a prioritized layer system (e.g., checking for High-Value metrics like Market Displacement or Monetization first) before tumbling down into broader buckets like Usability or Low Signal.
* **Results:** An entirely labeled, score-enriched DataFrame stored as a physical `.csv` asset containing macro-theme flags assigned to all 10,000+ customer records.

---

#### ⚙️ Script Blueprint & Data Routing
* **Execution Logic:** 
  1. Instantiates the sequential `categorize_review` conditional routing logic function.
  2. Creates a clean duplication slice of all positive records (`score == 5`).
  3. Invokes the `.apply()` method to pass text entries through the keyword matrix layer and creates the `final_theme` column.
  4. Counts and aggregates categorical repetitions across the entire dataset to evaluate volumes.
* **Output Format & Destination:** A comprehensive, multi-attribute tracking file exported directly to `paths.DATA_ANALYSIS_FULL_5STAR_CSV`, paired with an in-notebook terminal layout preview listing total counts sorted from highest to lowest volume (printed directly below the cell execution block).

```python
import pandas as pd

# 1. Load the full dataset (assuming 'df' is your main dataframe)
# We focus on 5-star reviews for the Value Proposition Analysis
pos_df = df[df['score'] == 5].copy()

# 2. Define the Waterfall Logic
def categorize_review(text):
    text = str(text).lower()
    
    # Priority 1: Market Displacement
    if any(word in text for word in ['better than', 'switched', 'replaced', 'competitor', 'other apps', 'others', 'many app']):
        return "Market Displacement"
    
    # Priority 2: Monetization
    elif any(word in text for word in ['premium', 'pro', 'subscription', 'subscribe', 'version', 'pay', 'worth it', 'free version', 'buy', 'allow', 'customizable', 'unlimited', 'tasks', 'advise', 'new', 'type', 'android', 'iphone', 'since', 'past']):
        return "Monetization Sentiment"
    
    # Priority 3: Retention
    elif any(word in text for word in ['day', 'year', 'month', 'daily', 'habit', 'every day', 'everyday', 'always', 'keep using', 'consistently', 'week', 'living without', 'keep track ']):
        return "Retention Drivers"
    
    # Priority 4: Efficiency
    elif any(word in text for word in ['mental', 'organized', 'save time', 'productive', 'life changer', 'focus', 'clear', 'life', 'better', 'fun', 'help', 'encourages', 'growth', 'increase', 'organized', 'manage', 'motivate', 'reminder', 'option', 'item', 'track', 'accountable', 'wanted', 'useful']):
        return "Efficiency"
    
    # Priority 5: Feature Requests
    elif any(word in text for word in ['wish', 'suggest', 'missing', 'feature', 'add', 'could', 'improvement', 'should', 'would', 'change', 'recommend', 'information']):
        return "Feature Requests & Friction"
    
    # Priority 6: Service Value & Trust
    elif any(word in text for word in ['support', 'responsive', 'fixed', 'thank you for fixing', 'reliable', 'trust', 'recommend']):
        return "Service Value & Trust"
    
    # Priority 7: User Workflows
    elif any(word in text for word in ['student', 'work', 'grocery', 'school', 'list', 'project', 'professional', 'listing', 'study', 'assignment', 'ADHD', 'sleeping', 'exercise', 'food', 'google calander']):
        return "User Archetypes & Workflows"
    
    # Priority 8: Usability (Functional)
    elif any(word in text for word in ['simple', 'clean', 'ui', 'easy', 'fast', 'design', 'interface', 'beautiful', 'smooth', 'gamified']):
        return "Product Usability & Aesthetics"
    
    # Layer 9: Catch-all for Vague Praise
    elif any(word in text for word in ['great', 'good', 'nice', 'awesome', 'love', 'best', 'amazing', 'love', 'excellent']):
        return "Low Signal"
    
    else:
        return "Noise"

# 3. Apply the engine to all 10,000+ rows
pos_df['final_theme'] = pos_df['content'].apply(categorize_review)

# 4. EXPORT: The Full Evidence File
# This is what you will use to build your final charts.
pos_df.to_csv(paths.DATA_ANALYSIS_FULL_5STAR_CSV, index=True)

# 5. PREVIEW: See the distribution of your themes
theme_counts = pos_df['final_theme'].value_counts()
print("✅ Categorization Complete. Here is your theme distribution:")
theme_counts
```

### Step 5: Strategic Synthesis: Engineering the Prioritized Value Proposition Matrix

* **What we are doing:** Aggregating the quantitative outputs of our classification engine, mapping executive-level business context to each discovered category, isolating high-fidelity representative user quotes, and generating a finalized, sorted strategic matrix report.
* **Why we are doing this:** Corporate stakeholders do not make strategic decisions using messy, raw data tables. To deliver immediate value to product managers and executives, we must synthesize our quantitative metric counts with qualitative human context into a single "Source of Truth." Sorting this matrix strictly by our operational priorities tells leadership exactly what the "crown jewels" of the application are, ensuring core value metrics are protected during upcoming release cycles.
* **How we are doing this:** We build explicit mapping dictionaries (`priority_map` and `context_map`) to structure our analytical lens. We then iterate through each thematic group, pull exact character-bound representative quotes from our dataset to preserve the authentic customer voice, calculate the volume weights, and apply a strict numeric sort across the finalized dataframe.
* **Results:** A high-level, presentation-ready Value Proposition canvas written directly to disk as a clean HTML report asset.

---

#### ⚙️ Script Blueprint & Data Routing
* **Execution Logic:** 
  1. Instantiates corporate priority hierarchies and operational definition maps to bind business context to raw category outputs.
  2. Aggregates row records from our classified dataset to compute total volume weights (`Count`) per theme.
  3. Deploys string-length filtering constraints (favoring reviews between 60 and 200 characters) to dynamically harvest high-flavor customer quotes.
  4. Formats, compiles, and forces a strict numerical sort based entirely on the priority hierarchy.
* **Output Format & Destination:** A standalone executive summary document written directly to disk at `paths.REPORTS_POS_VALUE_PROP_COMP_HTML`, supplemented by an immediate inline notebook rendering of the complete sorted matrix for deployment verification.

```python
import pandas as pd

# 1. Define the Priority and Context Maps
# We use 1-8 to avoid the ambiguity of 0.
priority_map = {
    "Market Displacement": 1,
    "Monetization Sentiment": 2,
    "Retention Drivers": 3,
    "Efficiency": 4,
    "Feature Requests & Friction": 5,
    "Service Value & Trust": 6,
    "User Archetypes & Workflows": 7,
    "Product Usability & Aesthetics": 8
}

context_map = {
    "Market Displacement": "Demonstrates successful acquisition of users from competing platforms.",
    "Monetization Sentiment": "Indicates high perceived value for premium features and subscription models.",
    "Retention Drivers": "Reflects the app's success in becoming a long-term, daily habit for users.",
    "Efficiency": "Highlights the emotional and cognitive benefits of organization and focus.",
    "Feature Requests & Friction": "Identifies opportunities for roadmap expansion even among satisfied users.",
    "Service Value & Trust": "Validates the reliability of the app and the effectiveness of support/fixes.",
    "User Archetypes & Workflows": "Maps specific use cases (work, school, life) to marketing personas.",
    "Product Usability & Aesthetics": "Confirms the core UI/UX performance and visual satisfaction."
}

# 2. Build the Synthesis Data
value_prop_list = []

# Iterate through the themes defined in our priority map
for theme, priority in priority_map.items():
    # Filter the pos_df for the specific theme
    subset = pos_df[pos_df['final_theme'] == theme]
    
    # We include the theme even if the count is low to maintain the priority list
    count = len(subset)
    
    # Find a representative quote
    if count > 0:
        # Prioritize quotes between 60 and 150 characters for better "flavor"
        quality_quotes = subset[(subset['content'].str.len() > 60) & (subset['content'].str.len() < 200)]
        if not quality_quotes.empty:
            quote = quality_quotes.iloc[0]['content']
        else:
            quote = subset.iloc[0]['content']
    else:
        quote = "N/A - No reviews met this criteria in this sample."

    value_prop_list.append({
        "Priority": priority,
        "Value Proposition": theme,
        "Context of Success": context_map.get(theme),
        "Count": count,
        "Evidence (User Voice)": f"\"{quote[:160]}...\""
    })

# 3. Create DataFrame and Final Sort
# We sort strictly by the 'Priority' column (1 to 8)
full_value_prop_table = pd.DataFrame(value_prop_list).sort_values(by='Priority')

# 4. EXPORT: Save for the Final Report
full_value_prop_table.to_html(paths.REPORTS_POS_VALUE_PROP_COMP_HTML, index=False)

print("✅ Success: Comprehensive Value Proposition Table (Sorted 1-8) saved.")
full_value_prop_table
```
