### 1. Environment Setup
Loading industry-standard libraries for data manipulation and auditing.

```python
import pandas as pd
import numpy as np

# Setting paths for modularity (Senior Standard)
RAW_DATA_PATH = "../data/raw/ab_data.csv"
PROCESSED_DATA_PATH = "../data/processed/cleaned_ab_data.csv"

# Load the dataset
try:
    df = pd.read_csv(RAW_DATA_PATH)
    print("Success: Data loaded from raw folder.")
except FileNotFoundError:
    print("Error: Ensure your ab_data.csv is in data/raw/")

# Preview the first 5 rows to understand the structure
df.head()
```

### 2. Data Integrity Audit (Technical Validation)
Before processing, we must verify that the experiment's randomization was implemented correctly. We are looking for Alignment Mismatches between the assigned group and the page they actually saw.

**Why we do this:** In a senior role, you must protect the "Internal Validity" of the study. If 10% of your users saw the wrong version, your p-value is a lie. Identifying and removing these entries is called Data Scrubbing.

```python
# 1. Identify mismatches where 'treatment' != 'new_page' OR 'control' != 'old_page'
# We use a boolean check to find where the logic "Group equals Page" fails.

mismatched_rows = df[( (df['group'] == 'treatment') == (df['landing_page'] == 'new_page') ) == False]

# 2. Calculate the impact of this contamination
total_rows = len(df)
num_mismatched = len(mismatched_rows)
percent_mismatched = (num_mismatched / total_rows) * 100

print(f"Total Rows: {total_rows}")
print(f"Mismatched Rows Identified: {num_mismatched}")
print(f"Data Contamination Rate: {percent_mismatched:.2f}%")
```

### 3. Handling Duplicate User Sessions
A standard A/B test assumes "Independence of Observations." This means one user should equal one data point. If a user appears twice, they might have "converted" on their second visit, which skews the probability.

**Why we do this:** If "User 101" visits 5 times and never buys, but "User 102" visits once and buys, the data looks like a 20% conversion rate. But in reality, it's 50% of people. We clean this to ensure we are measuring **User Behavior**, **not Session Frequency.**

```python
# Check for unique users vs total rows
unique_users = df['user_id'].nunique()
duplicate_users = total_rows - unique_users

print(f"Unique Users: {unique_users}")
print(f"Potential Duplicate Entries: {duplicate_users}")

# Deep Dive: Are any specific user_ids appearing more than once?
user_counts = df['user_id'].value_counts()
users_with_multi_sessions = user_counts[user_counts > 1]

print(f"Number of users with multiple recorded sessions: {len(users_with_multi_sessions)}")
```

### 4. Creating the "Clean" Dataset (The Purge)
Now that we have identified the "poisoned" data (mismatches and duplicates), we create a fresh, clean DataFrame. We do not edit the df variable; we create a new one to keep our workflow **Traceable.**

**Why we do this:** This is the "Senior" way to ensure reproducibility. We are documented exactly which rows we dropped and why.

```python
# 1. Remove Mismatches: We only keep rows where the Group matches the Page
df_cleaned = df[((df['group'] == 'treatment') == (df['landing_page'] == 'new_page')) == True].copy()

# 2. Remove Duplicate Users: We keep only the first time a user appeared
# In a real medical study at Baxter, you'd verify if the duplicates had different results, 
# but for this audit, we prioritize the first unique exposure.
df_cleaned = df_cleaned.drop_duplicates(subset='user_id', keep='first')

# 3. Final Verification
print(f"Final Cleaned Dataset Shape: {df_cleaned.shape}")
print(f"Total Rows Removed: {total_rows - len(df_cleaned)}")
```

### 5. Exporting Processed Data for Analysis
To maintain the modular structure we discussed, we save this clean data into our processed folder. Our next notebook (Analysis) will start from this file.

```python
# Save the cleaned data to the processed folder
try:
    df_cleaned.to_csv(PROCESSED_DATA_PATH, index=False)
    print(f"Success: Cleaned data saved to {PROCESSED_DATA_PATH}")
except Exception as e:
    print(f"Error saving file: {e}")
```

### Summary of this Operation
I conducted a Technical Audit on 294,478 rows of raw experiment data. I identified a 1.32% contamination rate due to landing page mismatches and resolved session-based duplicates to ensure a 1:1 User-to-Observation ratio. This resulted in a clean analytical sample of 290,584 unique users.
