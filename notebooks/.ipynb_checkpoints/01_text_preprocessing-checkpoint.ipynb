```python
import pandas as pd
import numpy as np
import re
import nltk
from nltk.corpus import stopwords

# Download the 'dictionary' of words to ignore (like 'the', 'is', 'at')
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

print("NLP Environment Ready.")
```

## 1. The Preprocessing Strategy (The "Senior" Way)
Before we let an AI model touch this, we have to "normalize" the text. Computers are bad at understanding that "Love it!!!", "love it", and "LOVE IT" all mean the same thing.

**Why we do this:** If we don't clean the text, our model will treat every variation as a different word, which "dilutes" the data. We need to:

**Lowercase everything.**

**Remove noise** (special characters, HTML links, and extra spaces).

**Handle Stopwords** (removing words like "the," "is," "at" that don't carry emotional meaning).

### 2. Loading and Auditing the Text
Let's load the data and check for "Empty Reviews." A review with a score but no text is useless for NLP.

```python
# Constants for pathing
#RAW_PATH = "../data/raw/google_play_reviews.csv"

# Load data - focusing on the columns we actually need for UX Research
df = pd.read_csv(paths.DATA_RAW_GOOGLE_REVIEWS_CSV, usecols=['content', 'score', 'thumbsUpCount', 'at'])

# Audit: How many reviews are missing text?
print(f"Total entries: {len(df)}")
null_counts = df['content'].isnull().sum()
print(f"Reviews missing text: {null_counts}")

# Drop rows where content is empty (Senior Protocol: Clean before processing)
df = df.dropna(subset=['content'])
```

### 3. The "Cleaning Machine" (A Reusable Function)
At a high level, you don't want to write cleaning code for every single row. You write a Function that can be applied to thousands of rows instantly.

```python
def clean_ux_text(text):
    """
    Standardizes review text for NLP analysis.
    - Lowers case
    - Removes special characters/digits
    - Removes stopwords
    """
    # 1. Lowercase
    text = str(text).lower()
    
    # 2. Remove special characters and digits using Regular Expressions (Regex)
    text = re.sub(r'[^a-z\s]', '', text)
    
    # 3. Tokenize (split into words) and remove stopwords
    words = text.split()
    cleaned_words = [w for w in words if w not in stop_words]
    
    return " ".join(cleaned_words)

# Apply the function to our content column
# Warning: This may take a minute if the dataset is 100k+ rows
print("Cleaning text... please wait.")
df['cleaned_content'] = df['content'].apply(clean_ux_text)

print("Success: Text normalization complete.")
df[['content', 'cleaned_content']].head()
```

### 4. Finish the Cleaning
Final cell to "Export" the work. This is the hand-off point.

```python
# Save the cleaned dataframe to the processed folder
# We keep 'content' for reference and 'score' for validation
#PROCESSED_PATH = "../data/processed/cleaned_reviews.csv"

df_to_save = df[['content', 'cleaned_content', 'score', 'at', 'thumbsUpCount']]
df_to_save.to_csv(paths.DATA_PROCESSED_CLEANED_CSV, index=False)

print(f"Data Cleaning Complete. File saved at: {paths.DATA_PROCESSED_CLEANED_CSV}")
```
