```python
import os
from pathlib import Path

def print_tree(directory, ignore_dirs=None):
    if ignore_dirs is None:
        # Added .DS_Store and config-related artifacts to the ignore list
        ignore_dirs = {'.ipynb_checkpoints', '.git', '__pycache__', '.virtual_documents', '.DS_Store'}
    
    # Use sorted() to keep the output organized alphabetically
    for path in sorted(Path(directory).rglob('*')):
        # Skip ignored directories and their contents
        if any(part in ignore_dirs for part in path.parts):
            continue
            
        # Calculate depth for indentation
        depth = len(path.relative_to(directory).parts)
        spacer = '    ' * (depth - 1)
        
        if path.is_dir():
            print(f"{spacer}📁 {path.name}/")
        else:
            print(f"{spacer}📄 {path.name}")

# --- THE FIX FOR NOTEBOOKS ---
# Path.cwd() gets '.../sentiment-analysis-audit/utils'
# .parent moves up to '.../sentiment-analysis-audit/' (Project Root)
root_directory = Path.cwd().parent

print(f"Mapping Project Root: {root_directory}\n")
print_tree(root_directory)
```
